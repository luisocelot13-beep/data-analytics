Duplicate the previous code cell, then customize the plot with a title (‘Example

```python
plt.plot(x,y,marker='o')
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
```

# Styling Plots

Plot the same data as above, but this time, change the line style to dashed ('--') and
the line color to red

```python
plt.plot(x, y, linestyle='--', color='red')
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
```

Change the marker style to a square ('s'), set the marker size to 10, and change the
marker color to green.

```python
plt.plot(x,y,linestyle='--',
         color='red',
         marker='s',
         markersize=10,
         markerfacecolor='green')
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
```

## Advanced Plotting Techniques

Create a bar chart showing the sales of different products. Use the following data:

```python
products = ['Product A', 'Product B', 'Product C', 'Product D']
sales = [23, 17, 35, 29]
```

```python
plt.bar(products,sales)
plt.title('Product Sales')
plt.xlabel('Products')
plt.ylabel('Sales')
```

Customize your bar chart by adding a title, x-axis label, and y-axis label. Set the
color of the bars to 'skyblue' and adjust the bar width to 0.5.

```python
plt.bar(products,sales,color='skyblue',width=0.5)
plt.title('Product Sales')
plt.xlabel('Products')
plt.ylabel('Sales')
```

Create a scatter plot to visualize the relationship between two variables. Use the
following data:
x = np.random.rand(50)
y = np.random.rand(50)

```python
import numpy as np
```

```python
x = np.random.rand(50)
y =np.random.rand(50)
plt.scatter(x,y)
plt.title('Scatter Plot Example')
plt.xlabel('X Values')
plt.ylabel('Y Values')
```

Customize your scatter plot by setting the size of the markers based on another
variable (e.g., s = np.random.rand(50) * 100), and using a colormap (e.g.,
cmap='viridis')

```python
x = np.random.rand(50)
y = np.random.rand(50)

# random sizes for markers
s = np.random.rand(50) * 100

#random colors for markers
colors = np.random.rand(50)

plt.scatter(x,y,s=s,c=colors,cmap='viridis')
plt.title('Customized Scatter Plot')
plt.xlabel('X Values')
plt.ylabel('Y Values')
```

# Subplots

Create a figure with two subplots: one for a line chart and one for a histogram

```python
# Data for sine wave
x = np.linspace(0,10,100)
y = np.sin(x)

#data for histogram
random_data = np.random.randn(1000)

#create figue with 2 subplots
fig,axs = plt.subplots(1,2, figsize=(12,5))

# first subplot: line chart
axs[0].plot(x,y)
axs[0].set_title('Sine Wave')
axs[0].set_xlabel('X Values')
axs[0].set_ylabel('sin(x)')

# Second subplot: histogram
axs[1].hist(random_data)
axs[1].set_title('Histogram of Random Numbers')
axs[1].set_xlabel('Value')
axs[1].set_ylabel('Frequency')
```
