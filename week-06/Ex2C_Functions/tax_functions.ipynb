# Ex2C functions
# In this lab we will create functions focused on specific tax codes,then create function with if,elif,else
# then  get withholdings depending on code.

```python
def get_soc_sec_tax(amount):
    total_tax = amount * 0.062
    
    return(
        f' Your total for ss tax is {total_tax} on the amount of {amount}')
```

```python
print(get_soc_sec_tax(700))
```

```python
def get_medicare_tax(amount):
    total_tax = amount * 0.0145
    return(
        f' Your medicare tax for the amount of {amount} is {total_tax}')
    
```

```python
print(get_medicare_tax(700))
```

```python
def get_federal_tax(amount,code):
    if code == 0:
        tax_rate = 0.23
    elif code == 1:
        tax_rate = 0.21
    elif code == 2:
        tax_rate = 0.195
    elif code == 3:
        tax_rate = 0.185
    else:
        extra_withholdings = code - 4
        tax_rate = 0.18 - (extra_withholdings * 0.005)
    hold_amount = amount * tax_rate
    return hold_amount
      
    
```

```python
print(get_federal_tax(100,1))
```

```python
print(get_federal_tax(300,2))
```

```python
print(get_federal_tax(500,3))
```
