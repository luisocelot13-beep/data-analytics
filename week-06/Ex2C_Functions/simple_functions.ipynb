# Working with defining custom functions

```python
def favorite_things(name,movie,music):
    return(
        f'My name is {name}\n'
        f'My favorite movie is {movie}\n'
        f'My favorite musican/band is {music}')
```

```python
print(favorite_things('Luis Lopez','american','Big'))
```

```python
def why_im_here(purpose):
    return(
        f'My main puropes is {purpose}')

        
```

```python
print(why_im_here('to make my mother happy'))
```

```python
def favorite_place(place,why):
    return(
        f'My favorite place in the world is {place} because {why}')
     
```

```python
print(favorite_place('Mexico','Its very hot'))
```

```python
print(favorite_place('India','Love the scenery'))
```
