# Working with Datetime functions

```python
import datetime
```

# Today

```python
today = datetime.datetime.now()
```

```python
print(today)
```

```python
print(today.strftime('Today is %A,%B %d, %Y '))
      
```

# My birthday 

```python
bday = datetime.date(1998,2,14)
```

```python
print(bday.strftime('my birthday is %A,%B %d,%Y'))
```

# Next 90 Days

```python
ninety_d = today + datetime.timedelta(days=90)
```

```python
print(ninety_d.strftime('90 days from today is %A, %B %d, %Y'))
```

# Time for Dinner

```python
dinner_time = datetime.time(6,30)
```

```python
print(dinner_time.strftime('Dinner is at %I:%M %p'))
```
