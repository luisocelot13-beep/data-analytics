# Exercise 2.A Using built in functions
## Lab 2. Parsing strings - part code

```python
def get_supplier_code(parts_code):
    for i in range(len(parts_code)):
        if parts_code[i] == ':':
            supply_code = parts_code[:i]
            return supply_code
            
```

```python
get_supplier_code('ACME:123-L')
```

```python
def get_product_num(parts_code):
    for i in range(len(parts_code)):
        if parts_code[i] == ':':
            start = i + 1
        elif parts_code[i] == '-':
            end = i
            product_num = parts_code[start:end]
            return product_num
            
```

```python
get_product_num('ACME:123-L')
```

```python
def get_size(parts_code):
    for i in range(len(parts_code)):
        if parts_code[i] == '-':
            size_i = parts_code[i + 1:]
            return size_i
    
    
```

```python
get_size('DI:12345-M')
```

```python
code = 'ACME:123-L'

```

```python
print(f' for {code}, the suppler is {get_supplier_code(code)}.'
      f' product  number is {get_product_num(code)} is available in size {get_size(code)}')
```
