## Lab 3. Employee data
# This lab exercis makes updates to information extracted from an employee database.

```python
hr_list = [('0123','HR','Rebecca Yang','69000'),
           ('0121','IT','Mark Blick','67000'),
           ('0068','IT','Kahna Larsen','112000'),
           ('0234','OPS','Jim Smith','54000')
           



]
```

```python
for i in range(len(hr_list)):
    if hr_list[i][2] == 'Mark Blick':
        hr_list[i] = ('0214','IT','Mark Blick Hawley','67000')
print(hr_list)
```

```python
for i in range(len(hr_list)):
    if hr_list[i][2] == 'Jim Smith':
        hr_list[i] = ('0214','CS','Jim Smith','6000')
```

```python
print(hr_list)
```

```python
for emp in hr_list:
    emp_id = emp[0]
    dept = emp[1]
    name = emp[2]
    salary = int(emp[3])

    print(f'{emp_id} | {dept} | {name} | $ {salary:.2f}')
```
