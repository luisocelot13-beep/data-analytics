## Lab 1. Working with names - parsing strings

```python
test_name = 'John Doe'
```

```python
def parse_and_display(name):
    for i in range(len(name)):
        if name[i] ==' ':
            first = name[:i]
            last = name [i+1:]
            print(f'Full name: {name}')
            print(f'First name: {first}')
            print(f'Last name: {last}')
```

```python
parse_and_display(test_name)
```

```python
test_list = ['Grace Flores','JB Oinonen','Ken Penn']
```

```python
parse_and_display(test_list[2])
```

```python
parse_and_display('Luis Lopes')
```

```python
def parse_and_display2(name):
    first_space = name.find(' ')
    second_space = name.find(' ',first_space + 1)
    first = name[:first_space]
    middle = name[first_space + 1:second_space:]
    last = name[second_space + 1:]

    print(f'full name: {name}')
    print(f'first name: {first}')
    print(f'Middle name: {middle}')
    print(f'Last name:{last}')
    
    
```

```python
parse_and_display2('Luis Armando lopez')
```
